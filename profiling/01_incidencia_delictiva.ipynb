# Profiling: Incidencia delictiva (SESNSP)
Solo Yucatán (clave de entidad 31).

```python
import io
import pandas as pd
from curl_cffi import requests
from ydata_profiling import ProfileReport

URL = "https://repodatos.atdt.gob.mx/api_update/sesnsp/incidencia_delictiva/IDM_NM_dic25.csv"
```

```python
r = requests.get(URL, impersonate="chrome", timeout=300)
r.raise_for_status()
df = pd.read_csv(io.BytesIO(r.content), encoding="latin1")
df = df[df["Clave_Ent"] == 31]
df.shape
```

```python
ProfileReport(df, title="Incidencia delictiva - Yucatán").to_file("01_incidencia_delictiva.html")
```
