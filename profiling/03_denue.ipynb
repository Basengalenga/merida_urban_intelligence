# Profiling: DENUE (INEGI), Yucatán
Un reporte HTML por edición.

```python
import io
import zipfile
import pandas as pd
import requests
from ydata_profiling import ProfileReport

BASE = "https://www.inegi.org.mx/contenidos/masiva/denue"
EDICIONES = {
    "2020_11": f"{BASE}/2020_11/denue_31_1120_csv.zip",
    "2021_11": f"{BASE}/2021_11/denue_31_1121_csv.zip",
    "2022_11": f"{BASE}/2022_11/denue_31_1122_csv.zip",
    "2023_11": f"{BASE}/2023_11/denue_31_1123_csv.zip",
    "2024_11": f"{BASE}/2024_11/denue_31_1124_csv.zip",
    "2025_05": f"{BASE}/2025_05/denue_31_0525_csv.zip",
}
```

```python
def descargar(url):
    r = requests.get(url, timeout=300)
    r.raise_for_status()
    partes = []
    with zipfile.ZipFile(io.BytesIO(r.content)) as z:
        for n in z.namelist():
            if n.lower().endswith(".csv") and "conjunto_de_datos" in n.lower():
                raw = z.read(n)
                try:
                    partes.append(pd.read_csv(io.BytesIO(raw), encoding="utf-8"))
                except UnicodeDecodeError:
                    partes.append(pd.read_csv(io.BytesIO(raw), encoding="latin-1"))
    return pd.concat(partes, ignore_index=True)
```

```python
for edicion, url in EDICIONES.items():
    df = descargar(url)
    print(edicion, df.shape)
    ProfileReport(df, title=f"DENUE {edicion} - Yucatán").to_file(f"03_denue_{edicion}.html")
```
