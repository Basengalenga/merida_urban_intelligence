# Profiling: SCIAN (INEGI), hoja CLASE
Versiones 2018 y 2023 en un solo reporte (columna `SCIAN`).

```python
import io
import pandas as pd
import requests
from ydata_profiling import ProfileReport

BASE = "https://www.inegi.org.mx/contenidos/app/scian"
URLS = {
    "SCIAN 2018": f"{BASE}/scian_2018_categorias_y_productos.xlsx",
    "SCIAN 2023": f"{BASE}/scian_2023_categorias_y_productos.xlsx",
}
```

```python
dfs = []
for version, url in URLS.items():
    r = requests.get(url, timeout=300)
    r.raise_for_status()
    # La fila 0 es el título del documento; los encabezados están en la fila 1
    d = pd.read_excel(io.BytesIO(r.content), sheet_name="CLASE", header=1).dropna(how="all")
    d["SCIAN"] = version
    dfs.append(d)
df = pd.concat(dfs, ignore_index=True)
df.shape
```

```python
ProfileReport(df, title="SCIAN - hoja CLASE").to_file("05_scian.html")
```
