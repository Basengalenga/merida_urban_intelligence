# Profiling: Marco Geoestadístico municipal (INEGI), Yucatán
Se excluye la columna `geometry` (ydata-profiling no perfila geometrías).

```python
import io
import tempfile
import zipfile
import geopandas as gpd
import requests
from ydata_profiling import ProfileReport

URL = "https://www.inegi.org.mx/contenidos/productos/prod_serv/contenidos/espanol/bvinegi/productos/geografia/marcogeo/889463807469/31_yucatan.zip"
```

```python
r = requests.get(URL, timeout=120)
r.raise_for_status()
with tempfile.TemporaryDirectory() as tmp:
    with zipfile.ZipFile(io.BytesIO(r.content)) as z:
        z.extractall(tmp)
        shp = next(n for n in z.namelist() if n.endswith("mun.shp"))
    mun = gpd.read_file(f"{tmp}/{shp}")
df = mun.drop(columns="geometry")
df.shape
```

```python
ProfileReport(df, title="Marco Geoestadístico municipal - Yucatán").to_file("04_marco_geo_municipal.html")
```
