# Profiling: ITER 2020 (INEGI), Yucatán

```python
import io
import zipfile
import pandas as pd
import requests
from ydata_profiling import ProfileReport

URL = "https://www.inegi.org.mx/contenidos/programas/ccpv/2020/datosabiertos/iter/iter_31_cpv2020_csv.zip"
```

```python
r = requests.get(URL, timeout=120)
r.raise_for_status()
with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    nombre = next(n for n in z.namelist() if n.lower().endswith(".csv") and "conjunto_de_datos" in n.lower())
    df = pd.read_csv(z.open(nombre))
df.shape
```

```python
ProfileReport(df, title="ITER 2020 - Yucatán", minimal=True, plot={"image_format": "png", "dpi": 50}).to_file("02_iter_2020.html")
```
